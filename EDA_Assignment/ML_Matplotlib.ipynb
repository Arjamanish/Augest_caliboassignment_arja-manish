```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv('student_marks.csv')
print(df)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 4.2))

ax.hist(df['Maths'], bins=10)

ax.set_title('Histogram of Maths Marks')
ax.set_xlabel('Maths Marks')
ax.set_ylabel('Number of Students')

plt.show()
```

```python
students = df['Student_Name'].head(10)
marks = df['Maths'].head(10)

fig, ax = plt.subplots(figsize=(7, 4.2))

ax.bar(students, marks)

ax.set_title('Bar Graph of Maths Marks for First 10 Students')
ax.set_xlabel('Student')
ax.set_ylabel('Maths Marks')

plt.xticks(rotation=45)
plt.show()
```

```python
days = list(range(1, 11))
avg = [62, 64, 65, 67, 70, 72, 75, 78, 80, 82]

fig, ax = plt.subplots(figsize=(7, 4.2))

ax.plot(
    days,
    avg,
    marker='o',
    color='green',
    linewidth=2,
    markersize=7
)

ax.set_title(' Line Graph of Average Marks Over 10 Tests')
ax.set_xlabel('Test Number')
ax.set_ylabel('Average Marks')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

ax.scatter(df['Attendance_Percentage'], df['Average_Marks'])

ax.set_title('Attendance vs Average Marks')
ax.set_xlabel('Attendance Percentage')
ax.set_ylabel('Average Marks')

plt.show()
```

```python
grade_counts = df['Grade'].value_counts()

fig, ax = plt.subplots(figsize=(7, 7))

ax.pie(
    grade_counts,
    labels=grade_counts.index,
    autopct='%1.1f%%'
)

ax.set_title('Student Grade Distribution')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
# (Make sure your CSV file is in the same folder and named 'covid_data.csv')
df = pd.read_csv('covid_data.csv')

# Convert the Date column so Python understands it's a date
df['Date'] = pd.to_datetime(df['Date'])

# ---------------------------------------------------------
# Task 1: Plot daily confirmed cases
# ---------------------------------------------------------
plt.figure(figsize=(8, 4))
sns.lineplot(data=df, x='Date', y='Confirmed', hue='Country')
plt.title('Daily Confirmed Cases by Country')
plt.show()

# ---------------------------------------------------------
# Task 2: Line plots for deaths and recoveries
# ---------------------------------------------------------
# Plotting Deaths
plt.figure(figsize=(8, 4))
sns.lineplot(data=df, x='Date', y='Deaths', hue='Country')
plt.title('Daily COVID-19 Deaths')
plt.show()

# Plotting Recoveries
plt.figure(figsize=(8, 4))
sns.lineplot(data=df, x='Date', y='Recovered', hue='Country')
plt.title('Daily COVID-19 Recoveries')
plt.show()

# ---------------------------------------------------------
# Task 3: Heatmap for country-wise statistics
# ---------------------------------------------------------
# Group the data by country and find the maximum number for each
country_stats = df.groupby('Country')[['Confirmed', 'Deaths', 'Recovered', 'Active']].max()

plt.figure(figsize=(7, 5))
sns.heatmap(country_stats, annot=True, fmt=".0f", cmap="Blues")
plt.title('Country-wise Statistics Heatmap')
plt.show()

# ---------------------------------------------------------
# Task 4: Analyze correlations
# ---------------------------------------------------------
plt.figure(figsize=(6, 4))
correlation = df[['Confirmed', 'Deaths', 'Recovered', 'Active']].corr()
sns.heatmap(correlation, annot=True, cmap="Reds")
plt.title('Correlation Matrix')
plt.show()

# ---------------------------------------------------------
# Task 5: Identify countries with the highest growth
# ---------------------------------------------------------
# Sort countries by their highest number of confirmed cases
highest_growth = df.groupby('Country')['Confirmed'].max().sort_values(ascending=False)

print("Countries ranked by highest confirmed cases:")
print(highest_growth)
```

```python
# ---------------------------------------------------------
# Task 3: Heatmap for country-wise statistics
# ---------------------------------------------------------
# Group by country and get the maximum recorded numbers
country_stats = df.groupby('Country')[['Confirmed', 'Deaths', 'Recovered', 'Active']].max()

# Make the figure a bit wider to match the reference image
plt.figure(figsize=(12, 6))

# cmap="Reds" makes it red. linewidths=0.5 adds the white grid lines.
sns.heatmap(country_stats, annot=True, fmt="d", cmap="Reds", linewidths=0.5, linecolor='white')

# Add the exact titles and labels from your image
plt.title('Country-wise COVID-19 Cases', fontsize=14)
plt.xlabel('COVID-19 Metrics')
plt.ylabel('Country')

plt.show()
```
